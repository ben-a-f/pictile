```python
## 1. Align images
from pictile import Tileset

# Create a Tileset from a directory containing photos
# of the same thing
tileset = Tileset("data")
# Align those images to the reference image, which is 
# the first image in the directory
tileset.align_sift()
# Crop the aligned images to they line up correctly
tileset.crop_aligned()
# Export all the aligned images
tileset.export_aligned("aligned")
```

```python
## 2. Make composite images
from pictile.tiles import stitch_tiles, export_png

# Split aligned images into a grid of equal tiles
tileset.make_tiles(3,3)

# Choose a couple of images
a = tileset.tiles[1][1]
b = tileset.tiles[2][1]

# Make a composite image, replacing the centre tile of 
# image a with the centre tile of image b
stiched_image = stitch_tiles([
    [a[0][0], a[0][1], a[0][2]],
    [a[1][0], b[1][1], a[1][2]],
    [a[2][0], a[2][1], a[2][2]],
])
export_png(stiched_image, "stitched//stitched.png")
```

```python
## 3. Create animations with built-in patterns
from pictile.animate import make_window_mp4, make_random_mp4

# Use our 3x3 tilesets to create an animation that replaces
# the centre tile of our reference image
window_imgs, window_durations = make_window_mp4(
    tileset,
    "animations//window.mp4",
    reverse=True
)

# Change to a 10x10 tile grid and create an animation that
# randomly replaces tiles
tileset.make_tiles(10,10)
rand_imgs, rand_durations = make_random_mp4(
    tileset,
    "animations//random.mp4",
    n=10,
    num_draws=48,
    duration=0.1,
    persist=1
)
```

```python
## 4. Define a custom animation using lists of images and durations
from pictile.animate import export_video

# Get two intact images from our tileset
va = tileset.aligned[1][1]
vb = tileset.aligned[2][1]

# Define the order in which images will be shown, 
# and the duration of each image
vid_images = [va, stiched_image, vb, stiched_image, 
              va, stiched_image, vb, stiched_image,
              va]
durations = [0.5] * len(vid_images)

# Create a custom animation
export_video(
    vid_images,
    durations,
    "animations//custom.mp4")
```
